```python
# Dictionaries


student = {
    "name": "Sara",
    "age": 20,
    "major": "Biotechnology"
}

print(student.get("name"), student.get("age"), student.get("GPA", "sth"))     # get (2 arguments)
student["GPA"] = 19                                                           # add
print(student.get("GPA"))
print()

# For juice, print not on the menu
# Print each item's price in order using => item()
# Print only the items cheaper than 3


menu = {"latte": 4.5, "tea": 2.0, "cake": 5.0}
print(menu.get("juice", "not on the menu"))
for item, price in menu.items():
    print(item, "costs", price)
print()

for item, price in menu.items():
    if price < 3:
        print(item)
```

```python
# Functions

def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Sara"))          # Hello, Sara!
print(greet("Ali", "Hi"))     # Hi, Ali!

# Create four functions of main operators, and use them in your calculator.

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b


num1 = float(input("Enter the first number: "))
operator = input("Enter an operator (+, -, *, /): ")
num2 = float(input("Enter the second number: "))

if operator == "+":
    print("Result:", add(num1, num2))
elif operator == "-":
    print("Result:", subtract(num1, num2))
elif operator == "*":
    print("Result:", multiply(num1, num2))
elif operator == "/":
    if num2 != 0:
        print("Result:", divide(num1, num2))
    else:
        print("Error: Cannot divide by zero!")
else:
    print("Invalid operator!")
```

```python
# Object Oriented Programming (OOP)

class student:
    def __init__(self, name, age, major):
        self.name = name
        self.age = age
        self.uni = "UT"
        self.major = major
        print("information saved")
    def greet(self):
        print("hello", self.name)
    def get_info(self):
        print(self.name, "is from", self.major)
stu1 = student("sara", 20, "biotech")
stu1.greet()
stu1.get_info()

# Create a cat class with name and age attributes and a Meow method.

class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def meow(self):
        print(f"{self.name} says Meow!")

cat1 = Cat("Luna", 3)

print(cat1.name)
print(cat1.age)
cat1.meow()
```
